Q2.4 : Each car station is managed by a "pole" at the Mairie of Toulouse. This pole of each station is given in the field `pole` of the field `fields` of each entry in the json file. Write a function `nbStationsOfPole` that returns a dictionary containting the number of stations managed by each pole.

```python
from collections import Counter
def nbStationsOfPole(data):
    stationsperpole = {}
    
    for stations in data["stations"]:
        
        pole = stations["fields"]["pole"]
        if pole not in stationsperpole:
            stationsperpole[pole] = 1
        else:
            stationsperpole[pole]+=1
    return stationsperpole

stationsperpole = nbStationsOfPole(data)
stationsperpole
```

Q2.6 : Add a new method `distanceTo` to the class CarStation. This method has two parameters lat and long (which correspond to the latitude and the longitude) and returns the distance (in kilometers) from the object to the point corresponding to the parameters. 

Note : We assume that the Earth is locally flat in Toulouse. Consider that 1 unity of latitude est equal to 110 km and 1 unity of longitude is equal to 80 km. 

```python
from math import sqrt
def distanceTo(self, pointlat, pointlong):
    latdif = abs(self.latitude - pointlat)*110
    longdif = abs(self.longitude - pointlong)*80
    distance = sqrt(latdif**2+longdif**2)
    return distance
CarStation.distanceTo = distanceTo

car_station.distanceTo(43.3, 3)
```

Q2.6 : Write a function `listCarStations` that returns a list containing the objects of type CarStation corresponding to the entries of the json.

```python
def listCarStations(data):
    car_stat_list =[]
    for entries in data["stations"]:
        car_stat_list.append(buildCarStation(entries))
    return car_stat_list
    
car_stat_list = listCarStations(data)
car_stat_list
```

Q2.7 : Write a function with 2 parameters (a latitude and a longitude) what returns the closest car station and the distance from the input location.

```python
def proximity(lat,long):
    distanceexit = []
    nmeexit = []
    for entry in car_stat_list:
        ind = car_stat_list.index(entry)
        distanceexit.append(entry.distanceTo(lat,long))
        nmeexit.append(car_stat_list[ind].name)
    
    return min(distanceexit), nmeexit[distanceexit.index(min(distanceexit))]

proximity(43.3,3)
```
